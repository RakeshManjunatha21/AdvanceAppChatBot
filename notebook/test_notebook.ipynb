```python
from google.ads.googleads.client import GoogleAdsClient
from google.ads.googleads.errors import GoogleAdsException
import pandas as pd
```

```python


# Load Google Ads Client (make sure google-ads.yaml is present or provide full path)
client = GoogleAdsClient.load_from_storage("google-ads.yaml")

# Your test customer ID (replace with actual)
customer_id = "1246259880"

# Test GAQL Query – change metrics/dimensions as needed
gaql_query = """
    SELECT
      campaign.id,
      campaign.name,
      metrics.clicks,
      metrics.impressions,
      metrics.average_cpc,
      segments.device
    FROM campaign
    WHERE segments.date DURING LAST_7_DAYS
    ORDER BY metrics.clicks DESC
    LIMIT 10
"""

# Function to run the GAQL and return results as DataFrame
def run_gaql_query(client, customer_id, query):
    ga_service = client.get_service("GoogleAdsService")
    results = []

    try:
        response = ga_service.search_stream(customer_id=customer_id, query=query)
        for batch in response:
            for row in batch.results:
                result_row = {}
                for field_name in row.__dict__.keys():
                    field_value = getattr(row, field_name)
                    result_row[field_name] = str(field_value)
                results.append(result_row)

        return pd.DataFrame(results)

    except GoogleAdsException as ex:
        print("Google Ads API Error:")
        for error in ex.failure.errors:
            print(f" - {error.message}")
        return pd.DataFrame()

# Run it
df = run_gaql_query(client, customer_id, gaql_query)

# Display the result
df.head()
```

```python
from google.ads.googleads.client import GoogleAdsClient
from google.ads.googleads.errors import GoogleAdsException

# Load the config (google-ads.yaml)
client = GoogleAdsClient.load_from_storage("google-ads.yaml")

def list_accessible_customers():
    try:
        customer_service = client.get_service("CustomerService")
        response = customer_service.list_accessible_customers()

        print("Accessible customer resource names (IDs):")
        for resource_name in response.resource_names:
            # Example format: customers/1234567890
            customer_id = resource_name.split("/")[1]
            print(" -", customer_id)
    except GoogleAdsException as ex:
        print("Google Ads API error:")
        for error in ex.failure.errors:
            print(f" - {error.message}")

# Call the function
list_accessible_customers()
```
